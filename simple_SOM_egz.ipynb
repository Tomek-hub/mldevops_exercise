```python
from minisom import MiniSom
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img = plt.imread('house_img.png')
```

```python
plt.imshow(img)
```

```python
img.shape
```

```python
pixels = np.reshape(img, (img.shape[0]*img.shape[1], 4))
```

```python
pixels.shape
```

```python
some = MiniSom(x= 4, y = 4, input_len = 4, sigma=0.1, learning_rate=0.2)
```

```python
some.random_weights_init(pixels)
```

```python
starting_weights = some.get_weights().copy()
```

```python
some.train_random(pixels, 100)
```

```python
qnt = some.quantization(pixels)
```

```python
clustered = np.zeros(img.shape)
```

```python
for i, q in enumerate(qnt):
  clustered[np.unravel_index(i, (img.shape[0], img.shape[1]))] = q
```

```python
plt.figure(figsize=(12, 6))
plt.subplot(221)
plt.title('Original')
plt.imshow(img)
plt.subplot(222)
plt.title('Result')
plt.imshow(clustered)
```

```python
plt.subplot(223)
plt.title('Initial Colors')
plt.imshow(starting_weights)
plt.subplot(224)
plt.title('Learnt Colors')
plt.imshow(some.get_weights()) # Bloczek 0,0 robi różnicę
```
